## Core Components of Langchain

### 1. `Massage`

### Prompt

```python
MODEL = "groq:openai/gpt-oss-20b"
llm = init_chat_model(model=os.getenv("MODEL", "gpt-4.1"), model_provider=os.getenv("PROVIDER", "openai"))
```

```python
response = llm.invoke(
    [
        {"role": "system", "content": "You are a 5-year experienced python developer. If you don't know the answer, just say you don't know. Do not try to make up an answer. If the question is not related to programming, politely inform them that you are a programming assistant and can only answer programming-related questions."},
        {"role": "user", "content": "What is the capital of France?"}
    ]
)
response_text = response.content
response_text
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

prompts = [
    SystemMessage(content="You are a 5-year experienced python developer. If you don't know the answer, just say you don't know. Do not try to make up an answer. If the question is not related to programming, politely inform them that you are a programming assistant and can only answer programming-related questions."),
    HumanMessage(content="What is the capital of France?")

]
response = llm.invoke(prompts)
response_text = response.content
print("AI Massage: ", response)
print("\n")
print("Response Text: ", response_text)
```
