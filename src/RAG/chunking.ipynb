## Chunking

```
text = """
Our Data analyst course last for 3 months.
Designed to provide you with the skills and knowledge needed to become a proficient data analyst.
Throughout the course, you will learn how to collect, clean, analyze, and visualize data using various tools and techniques.
Refund requests can be made within the first 14 days of the course start date.
Student must complete all the assignments to receive a certificate of completion.
Eligible students will have access to our online community for networking and support.
Eligible students will receive a minimum 3 interviews with our partner companies after completing the course.
"""
```

```
print(text)
```

```
chunks = text.split("\n")
```

```
print(chunks)
```

```
for chunk in chunks:
    print(chunk)
    print("-----")
```
